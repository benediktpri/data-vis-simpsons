```python
import pandas as pd
import altair as alt
alt.data_transformers.enable('json')
```

```python
df_episodes = pd.read_csv("C:\\Users\\20203666\\Documents\\GitHub\\data-vis-simpsons\\assignment2\\data\\raw_data\\simpsons_episodes.csv")
```

```python
df_lines = pd.read_csv("C:\\Users\\20203666\\Documents\\GitHub\\data-vis-simpsons\\assignment2\\data\\raw_data\\simpsons_script_lines.csv")
df_lines = df_lines.merge(
    df_episodes[['id', 'number_in_season', 'season']],
    left_on='episode_id',
    right_on='id',
    how='left'
)
df_lines['word_count'] = pd.to_numeric(df_lines['word_count'], errors='coerce')
df_lines = df_lines[df_lines['word_count'] <= 150]
df_lines.dropna(subset=['raw_character_text', 'normalized_text', 'word_count'], inplace=True)
df_lines.drop(['id_x', 'id_y'], axis=1, inplace=True)
df_lines
```

```python
df_lines[df_lines['raw_character_text'].str.startswith('Bart ', na=False)]
```

```python
top_characters = df_lines['raw_character_text'].value_counts().head(100).index.tolist()
char_to_del = ['Crowd', 'Narrator', 'Kids', 'Man']
char_to_rep = {'TROY': 'Troy McClure', 'DOLPH': 'Dolph', 'HERB': 'Herb', 'BUCK': 'Buck'}
top_characters
```

```python
df_q1 = df_lines.groupby(['raw_character_text', 'episode_id'])['word_count'].sum().reset_index()
df_q1 = df_q1[df_q1['raw_character_text'].isin(top_characters)]
df_q1
```

```python
char_dropdown = alt.binding_select(options=top_characters, name='Character: ')
char_select = alt.selection_point(fields=['raw_character_text'], bind=char_dropdown, value=top_characters[0])

chart = alt.Chart(df_q1).mark_bar(
    size=12 # Optional: adjust bar width for readability
).encode(
    x=alt.X('episode_id:Q', title='Episode Number'),
    y=alt.Y('word_count:Q', title='Words Spoken'), # Use directly, no aggregate
    color=alt.Color('raw_character_text:N', legend=None),
    tooltip=['episode_id', 'word_count']
).add_params(
    char_select
).transform_filter(
    char_select
).properties(
    width=600,
    height=300,
    title="Character Dialogue Volume per Episode"
)

chart.show()
```
