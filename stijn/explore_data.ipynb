```python
import pandas as pd
import altair as alt
```

```python
df_episodes = pd.read_csv("../data/kaggle_data/simpsons_episodes.csv")
df_episodes
```

```python
df_episodes = df_episodes[df_episodes['season'] != 28]
df_episodes
```

## fill in missing values
found at https://en.wikipedia.org/wiki/List_of_The_Simpsons_episodes_(seasons_1%E2%80%9320)


```python
df_episodes[(df_episodes['season']==8) & (df_episodes['number_in_season']==7)]['us_viewers_in_millions'] = 12.2
df_episodes[(df_episodes['season']==8) & (df_episodes['number_in_season']==8)]['us_viewers_in_millions'] = 14.4
df_episodes[(df_episodes['season']==8) & (df_episodes['number_in_season']==20)]['us_viewers_in_millions'] = 13.3
df_episodes
```

```python
season_years = df_episodes.groupby('season')['original_air_year'].agg(['min', 'max'])

season_years['season_label'] = (
        'Season ' + season_years.index.astype(str) +
        ' (' + season_years['min'].astype(str) + ')'
        #'-' + season_years['max'].astype(str) + ')'
)

df_episodes['season_label'] = df_episodes['season'].map(season_years['season_label'])
df_episodes
```

## Question 1

```python
df_ratings = df_episodes[['original_air_year', 'imdb_rating', 'season']]
df_ratings = df_ratings.groupby('season').agg({'imdb_rating': 'mean','original_air_year': 'median'})
df_ratings['original_air_year'] = round(df_ratings['original_air_year']+0.1)
df_ratings
```

```python
alt.Chart(df_ratings).mark_line(point=True).encode(
    x=alt.X('original_air_year:O', title='Year'),
    y=alt.Y('imdb_rating:Q', title='Rating')
)
```

```python
df_ratings = df_episodes[['season_label', 'imdb_rating', 'season']]
df_ratings = df_ratings.groupby('season_label').agg({'imdb_rating': 'mean', 'season': 'median'})
df_ratings['season_label'] = df_ratings.index
df_ratings
```

```python
alt.Chart(df_ratings).mark_line(point=True).encode(
    x=alt.X('season_label:O', title='Year', sort=alt.SortField(field='season', order='ascending')),
    y=alt.Y('imdb_rating:Q', title='Rating', scale=alt.Scale(domain=[6, 9]))
)
```

```python
df_ratings = df_episodes[['season_label', 'imdb_rating', 'season', 'number_in_season']]
alt.Chart(df_ratings).mark_rect().encode(
    x=alt.X('number_in_season:O', title='Episode Number'),
    y=alt.Y('season_label:O', title='Season', sort=alt.SortField(field='season', order='ascending')),
    color=alt.Color('imdb_rating:Q', scale=alt.Scale(scheme='goldgreen'), title='IMDb Rating'),
    tooltip=['imdb_rating']
).properties(
    title='The Simpsons IMDb Ratings Heatmap',
    width=600,
    height=400
).configure_axis(
    labelFontSize=12,
    titleFontSize=14
)
```

## Question 2

```python
df_viewers = df_episodes[['original_air_year', 'us_viewers_in_millions', 'season']]
df_viewers = df_viewers.groupby('season').agg({'us_viewers_in_millions': 'mean', 'original_air_year': 'median'})
df_viewers['original_air_year'] = round(df_viewers['original_air_year'] + 0.1)
```

```python
alt.Chart(df_viewers).mark_line(point=True).encode(
    x=alt.X('original_air_year:O', title='Year'),
    y=alt.Y('us_viewers_in_millions:Q', title='Viewers in millions')
)
```

```python
df_viewers = df_episodes[['original_air_year', 'us_viewers_in_millions', 'season']]
df_viewers = df_viewers.groupby('original_air_year').agg({'us_viewers_in_millions': 'mean', 'original_air_year': 'median'})
alt.Chart(df_viewers).mark_line(point=True).encode(
    x=alt.X('original_air_year:O', title='Year'),
    y=alt.Y('us_viewers_in_millions:Q', title='Viewers in millions')
)
```

```python
df_viewers = df_episodes[['original_air_date', 'us_viewers_in_millions']]
df_viewers['date'] = pd.to_datetime(df_viewers['original_air_date'])
alt.Chart(df_viewers).mark_line(point=True).encode(
    x=alt.X('date:T', title='Date'),
    y=alt.Y('us_viewers_in_millions:Q', title='Viewers in millions')
)
```

```python
# 1. Prepare the data
df_viewers = df_episodes[['original_air_date', 'us_viewers_in_millions']].copy()
df_viewers['date'] = pd.to_datetime(df_viewers['original_air_date'])


scatter = alt.Chart(df_viewers).mark_circle(opacity=0.5, size=40, color='steelblue').encode(
    x=alt.X('date:T', title='Air Date'),
    y=alt.Y('us_viewers_in_millions:Q', title='Viewers (Millions)')
)


line = alt.Chart(df_viewers).mark_line(color='red', size=2, point=False).encode(
    x='year(date):T',
    y='mean(us_viewers_in_millions):Q'
)
chart = scatter + line
chart
```

## Question 3

```python
df_corr = df_episodes[['imdb_rating', 'us_viewers_in_millions']]
```

```python
scatter = alt.Chart(df_corr, title='Correlation Between Viewers and Gradings').mark_circle(opacity=0.5).encode(
    x=alt.X('us_viewers_in_millions', title='Viewers (Millions)', scale=alt.Scale(zero=False)),
    y=alt.Y('imdb_rating', title='Grading (1-10)', scale=alt.Scale(zero=False))
)
```

```python
chart = scatter + scatter.transform_regression('us_viewers_in_millions', 'imdb_rating').mark_line(color='red')
chart
```
